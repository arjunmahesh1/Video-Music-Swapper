## Step 4: Set Spotify Credentials

1. Click 🔑 icon in left sidebar
2. Add secrets:
   - `SPOTIPY_CLIENT_ID`

```python
# Option B: Load from Colab Secrets
from google.colab import userdata
import os

try:
    os.environ['SPOTIPY_CLIENT_ID'] = userdata.get('SPOTIPY_CLIENT_ID')
    os.environ['SPOTIPY_CLIENT_SECRET'] = userdata.get('SPOTIPY_CLIENT_SECRET')
    os.environ['SPOTIPY_REDIRECT_URI'] = 'http://localhost:8501'
    
    # Create .env file
    with open('.env', 'w') as f:
        f.write(f"SPOTIPY_CLIENT_ID={os.environ['SPOTIPY_CLIENT_ID']}\n")
        f.write(f"SPOTIPY_CLIENT_SECRET={os.environ['SPOTIPY_CLIENT_SECRET']}\n")
        f.write(f"SPOTIPY_REDIRECT_URI={os.environ['SPOTIPY_REDIRECT_URI']}\n")
    
    print("✅ Credentials loaded from Secrets")
except:
    print("⚠️  No secrets found. Make sure .env file was uploaded or add secrets.")
```

```python
import subprocess
import time
from pyngrok import ngrok

# Kill any existing Streamlit processes
!pkill -f streamlit
time.sleep(2)

# Start Streamlit in background
process = subprocess.Popen(
    ["streamlit", "run", "app.py", "--server.port=8501"],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE
)

# Wait for Streamlit to start
time.sleep(5)

# Create public URL with ngrok
public_url = ngrok.connect(8501)

print("\n" + "="*60)
print("🎵 VIDEO MUSIC SWAPPER IS RUNNING!")
print("="*60)
print(f"\n🌐 Public URL: {public_url}")
print(f"\n⚡ GPU Accelerated: {torch.cuda.is_available()}")
print("\n📌 Keep this cell running. Stop it to shut down the app.")
print("="*60)

# Keep running
try:
    process.wait()
except KeyboardInterrupt:
    print("\n🛑 Shutting down...")
    process.terminate()
```

## Troubleshooting

**App won't start:**
- Make sure all files were uploaded
- Check `.env` file exists with valid credentials

**No GPU detected:**
- Go to Runtime → Change runtime type → Select T4 GPU

**ngrok error:**
- Make sure you pasted your auth token correctly
- Get free token from ngrok.com

**Spotify authentication:**
- The OAuth redirect might not work perfectly in Colab
- You may need to manually copy the redirect URL
